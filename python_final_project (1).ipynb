```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python

df = pd.read_csv(r'C:\Users\Hp\Downloads\Diwali Sales Data.csv', encoding='unicode_escape')
```

```python
df.shape
```

```python
df.head(10)
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis=1,inplace= True) # drop unrelated/blank columns
```

```python
# we want to check how many nul values are there in each column
pd.isnull(df).sum()
```

```python
df.dropna(inplace=True)
# we want to drop the null values
```

```python
df.shape
```

```python
# We want to change the data type from string to int
df['Amount']=df['Amount'].astype('int')
```

```python
df['Amount'].dtypes
```

```python
#we want to check the info of only some specific columns
df[['Age','Orders','Amount']].describe()
```

```python
# WE WILL BE NOW DOING EXPLORATORY DATA ANALYSIS
```

```python
ax=sns.countplot(x='Gender',data=df)
for bars in ax.containers:
    ax.bar_label(bars)
    # plotting a bar chart for Gender and it's count
```

```python
sales_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data=sales_gen)
sns.set(rc={'figure.figsize':(10,5)})
# plotting a bar chart for gender vs total amount
```

```python
df.columns
```

```python
#Number of products vs age group
ax=sns.countplot(data=df,x='Age Group',hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
#Total amount vs age group
sales_age=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Age Group',y='Amount',data=sales_age)
```

```python
#From above graphs we can see that most of the buyers are of age group between 26-35 yrs female
```

```python

# total number of orders from top 10 states
sales_state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10) #shows only top 10 states
sns.set(rc={'figure.figsize':(15,5)}) # sets the size of the figure
sns.barplot(data=sales_state,x='State',y='Orders',hue='State')
```

```python
#total amount of sales from the top 10states
sales_state=df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(25,5)})
sns.barplot(data=sales_state,x='State',y='Amount',hue='State')
```

```python
#Even though the count is more for Kerela yet Haryana spends more in amount
```

```python
sns.set(rc={'figure.figsize':(30,5)})
ax=sns.countplot(data=df,x='Product_Category',hue='Product_Category')
for bars in ax.containers:
    ax.bar_label(bars)
    
```

```python
print('From above graphs we can see that most of the sold products are from Food, Clothing and Electronics category')
```

```python
sales_state = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_Category',y= 'Amount',hue='Product_Category')
```

```python
print("People spend the most on food")
```

```python
sales_state = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_ID',y= 'Orders')
```

```python
 print("top 10 most sold products")
```

```python
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(data = df, x = 'Occupation')

for bars in ax.containers:
    ax.bar_label(bars) # IT Sector buys the most products
```

```python
sales_state = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Occupation',y= 'Amount')
```

```python
print("From above graphs we can see that most of spending is by the buyers  working in IT, Healthcare and Aviation sector")
```
